```python
import numpy as np

X = np.array([[2.54, 4.33, 3.99, 2.57], # 환자 0
[3.12, 3.87, 3.84, 3.04], # 환자 1
[2.76, 4.17, 5.63, 3.28], # 환자 2
[3.87, 3.56, 4.25, 3.65], # 환자 3
[3.55, 3.91, 2.68, 4.22], # 환자 4
[4.12, 2.86, 3.30, 3.71]]) # 환자 5

labels = np.array(["정상", "정상", "정상", "확진", "확진", "확진"])

X
```

```python
new = np.array([[3.24, 3.68, 3.82, 3.77]])
new
```

```python
dist = np.sum(np.sqrt((X - new)**2), axis=1)
```

```python
who = dist.argmin() # 최솟값인 인덱스 반환
```

```python
labels[who]
```

```python
from sklearn.neighbors import KNeighborsClassifier

knc = KNeighborsClassifier(n_neighbors=1)
knc.fit(X, labels) # 거리 계산하지 않음, 구조만 맞추고 매개변수 넣는 등 준비만 함
```

```python
knc.predict(new)
```

```python
# k = 3
k = 3
temp = np.argsort(dist)
idx = temp[:k]
idx
```

```python
pred = labels[idx]
pred
```

```python
np.sum(labels[idx] == '정상')
```

```python
np.sum(labels[idx] == '확진')
```

```python
if(np.sum(pred == '확진') > np.sum(pred == '정상')): print('확진')
else: print('정상')
```

```python
from sklearn.neighbors import KNeighborsClassifier

knc = KNeighborsClassifier(n_neighbors=3)
knc.fit(X, labels) # 거리 계산하지 않음, 구조만 맞추고 매개변수 넣는 등 준비만 함
```

```python
# 입력 데이터 (다른 과목 점수)
X = np.array([ [7.5, 7.5, 7.0, 9.5, 8.5],# 학생0
[7.5, 7.0, 7.5, 8.0, 8.0], # 학생1
[8.0, 7.0, 8.0, 8.0, 8.5], # 학생2
[8.5, 8.0, 9.5, 7.5, 6.0], # 학생3
[10.0, 9.5, 9.0, 7.5, 7.5], # 학생4
[9.0, 9.0, 8.0, 8.0, 8.0]]) # 학생5
# 정답 (인공지능 점수)
y = np.array([5.0, 6.0, 8.5, 7.0, 10.0, 9.0])
# 신입 학생
new = np.array([9.0, 8.5, 8.0, 7.0, 8.0])
```

```python
dist = np.sqrt(np.sum((X - new)**2, axis=1))
dist
```

```python
idx = np.argsort(dist)
idx
```

```python
k = 3
y[idx[:k]]
```

```python
np.mean(y[idx[:k]])
```

```python
from sklearn.neighbors import KNeighborsRegressor
knr = KNeighborsRegressor(n_neighbors=k)
knr.fit(X, y) # 유클리드 거리 계산하지 않음, 준비만 함
```

```python
knr.predict([new])
```

```python
import pandas as pd

test = pd.read_csv("./data/basketball_test.csv")
train = pd.read_csv("./data/basketball_train.csv")

print(test.head())
print(train.head())
```

```python
X_tr = train[["3P", "BLK"]].values # 값만 추출하여 넘파이 배열 객체로 변환
Y_tr = train["Pos"].values

X_te = test[["3P", "BLK"]].values
Y_te = test["Pos"].values

X_tr[:5], Y_tr[:5], X_te[:5], Y_te[:5] 
```

```python
X_tr.shape
```

```python
from sklearn.metrics import accuracy_score
k = 1 # 1 ~ 10까지 넘파이 배열 생성
k
```

```python
scores = []

for k in range(1, 11):
    m = KNeighborsClassifier(n_neighbors=k)
    m.fit(X_tr, Y_tr) # 유클리드 거리 계산 하지 않음
    y_hat = m.predict(X_te)
    scores.append(accuracy_score(Y_te, y_hat))
scores
```

```python
np.argmax(scores)+1 # argmax: 큰 값 중에 가장 앞 쪽 인덱스 반환
```

```python
k=3
m = KNeighborsClassifier(n_neighbors=k)
m.fit(X_tr, Y_tr) # 유클리드 거리 계산 하지 않음
y_hat = m.predict(X_te)
```

```python
from sklearn.metrics import confusion_matrix # 실제값 먼저, 예측값 다음
from sklearn.metrics import classification_report

# m.classes_ : 분류한 클래스 이름 반환
# 0 : 'C'
# 1 : 'SG'
# [0, 0] : 실제값 'C'를 'C'로 예측
# [0, 1] : 실제값 'C'를 'SG'로 예측
# [1, 0] : 
# [1, 1] : 
confusion_matrix(Y_te, y_hat)
```

```python
m.classes_
```

```python
print(classification_report(Y_te, y_hat))
```
